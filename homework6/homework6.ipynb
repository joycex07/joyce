```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    if n == 0 or n == 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

def row_prime(array):
    result = []
    for row in array:
        if any(is_prime(x) for x in row):
            result.append(row)
    return np.array(result)

print(row_prime(arr))
```

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    
    for i in range(8):
        board[i, i % 2::2] = 1
        
    return board
```

```python
board_array = checkerboard()
print(board_array)
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    
    for i in range(8):
        board[i, (i + 1) % 2::2] = 1
        
    return board
```

```python
reverse_array = reverse_checkerboard()
print(reverse_array)
```

```python
def expansion(arr, spaces):
    result = np.empty(arr.shape, dtype=object)
    i = 0
    while i < len(arr):
        word = arr[i]
        expanded = ""
        j = 0
        while j < len(word):
            expanded += word[j]
            if j != len(word) - 1:
                k = 0
                while k < spaces:
                    expanded += " "
                    k += 1
            j += 1
        result[i] = expanded
        i += 1
    return result

universe = np.array(['galaxy', 'clusters'])
expansion(universe, 1)
```
