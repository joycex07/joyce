```python
# All values are type strings, and storing numeric values as strings is a problem for data analysis because you cannot apply those functions to the data if they are not type integers or floats.
```

```python
for col in df.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
```

```python
# The three different data types in the df are object, float64, and int64.
# Float64: decimal
# Int64: whole numbers
# Object: strings
# Planet, Ring System, and Global Magnetic Field are still object because we did not change those data types when using the for loop.
```

```python
def attach_units(col):
    if col == "Mass (10^24kg)":
        unit = 1e24 * u.kg
        new_col = "Mass (kg)"
        df[col] = [val * unit for val in df[col]]
        df.rename(columns={col: new_col}, inplace=True)
    elif col == "Diameter (km)":
        df[col] = [val * u.km for val in df[col]]
    elif col == "Density (kg/m^3)":
        df[col] = [val * (u.kg / u.m**3) for val in df[col]]
    elif col == "Gravity (m/s^2)":
        df[col] = [val * (u.m / u.s**2) for val in df[col]]
    elif col == "Escape Velocity (km/s)":
        df[col] = [val * (u.km / u.s) for val in df[col]]
    elif any(word in col for word in ["Distance", "Perihelion", "Aphelion"]):
        unit = 1e6 * u.km
        df[col] = [val * unit for val in df[col]]
        new_name = col.replace("(10^6 km)", "(km)")
        df.rename(columns={col: new_name}, inplace=True)
```

```python
for col in list(df.columns):
    attach_units(col)
```

```python
semi_major_axis = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2
```

```python
df.insert(df.columns.get_loc("Aphelion (km)") + 1, "Semi-Major Axis (km)", semi_major_axis)
df.columns
```

```python
df["Orbital Period (days)"] = df["Orbital Period (days)"] * u.day
df["Orbital Period (years)"] = df["Orbital Period (days)"].apply(lambda x: (x * u.day).to(u.year))
```

```python
df.drop(columns = ["Orbital Period (days)"], inplace=True)
df.columns
```

```python
period = df.loc[df["Planet"] == "Pluto", "Orbital Period (years)"].iloc[0]
print(f"{"Pluto"}'s orbital period is {period.value:.4f} years.")
```

```python
distance_cols = ["Distance from Sun (km)", "Perihelion (km)", "Aphelion (km)", "Semi-Major Axis (km)"]
```

```python
for col in distance_cols:
    new_col = col.replace("(km)", "(AU)") 
    df[new_col] = df[col].apply(lambda x: x.to(u.AU))
    df.drop(columns = [col], inplace=True)
```

```python
for col in ["Distance from Sun (AU)", "Perihelion (AU)", "Aphelion (AU)", "Semi-Major Axis (AU)"]:
    value = df.loc[df["Planet"] == "Pluto", col].iloc[0]
    print(f"Pluto {col}: {value:.4f}")
```
